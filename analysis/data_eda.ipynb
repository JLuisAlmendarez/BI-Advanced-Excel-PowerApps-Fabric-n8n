### Observaciones sobre cada dataframe:

__Customer__: Contiene PII del cliente. Hay un total de ```99,441 clientes```. Existen dos identificadores ```customer_unique_id``` para el cliente y ```customer_id```para cada orden que haga el cliente.
<br>
__Geolocations__: Es la base de datos de geografica de los poligonos o colonias asociadas a un codigo postal a los que la plataforma da servicio hay un total de ```1,000,163 ubicaciones```.
<br>
__Orders__: Contiene informacion sobre el status de la orden, timestamps relacionadas aprobacion de la compra yal  transporte del producto. En el momento hay un total de ```99,441 ordenes```. La mayoria entregadas, algunas con status pendiente.
<br>
__Products__: Contienen caracteristicas del producto y la categoria que pertenecen. Hay un total de ```32951 productos``` hasta ahora.
<br>
__Order Items__: Contiene informacion de los productos asociados con las ordenes, es la capa intermedia entre los productos disponibles, el vendedor, las ordenes y el cliente. Ademas incluye su precio, su valor de carga y el limite de entrega actualmente hay  ```112650 items siendo procesados```.
<br>
__Order Payments__: Contiene informacion sobre el metodo de pago, el valor de la orden, el numero de transacciones y el numero de meses de financiamiento para cada orden. ```103,886 pagos de ordenes``` con la informacion de sus financiamientos.
<br>
__Order reviews__: Contiene la informacion de la reseñas o calificaciones. Ademas esta asociada a las ordenes directamente, no a cada item individual. Existen ```99,224 calificaciones``` y ```40977 reseñas```.
<br>
__Sellers__: Es la lista de los vendedores, contiene PII del vendedor. Esta asociada a cada uno de los items que se venden. En total son ```3,095 vendedores```.
<br>
__Categories__: Es la traduccion al ingles de las categorias de los productos en total son ```71 categorias```.
